# Projective jet high-order check

This notebook imports `Jet_test.py` and checks the projective jet construction against the direct Riccati hierarchy.

The imported file uses exact rational arithmetic with `Fraction`. This is mathematically clean but can become slow at very high order. The default below is `MAX_T = 1000`. If it is too slow on your machine, reduce `TRIALS` to 1 or lower `MAX_T`.

```python
# Parameters
MAX_T = 1000      # Highest tensor-running order tested: T_2, ..., T_MAX_T.
TRIALS = 1        # Increase for more random rational trials.
SEED = 4321
VERBOSE = True
```

```python
from pathlib import Path
import importlib.util
import time

module_path = Path('Jet_test.py')
if not module_path.exists():
    module_path = Path('/mnt/data/Jet_test.py')

spec = importlib.util.spec_from_file_location('Jet_test', module_path)
jet = importlib.util.module_from_spec(spec)
spec.loader.exec_module(jet)

print(f'Loaded: {module_path.resolve()}')
```

```python
start = time.time()
status = jet.projective_jet_check(
    max_index=MAX_T,
    trials=TRIALS,
    seed=SEED,
    verbose=VERBOSE,
)
elapsed = time.time() - start

jet_ok = all(status.values())
failed = [i for i, ok in status.items() if not ok]

print()
print(f'Projective jet check through T_{MAX_T}: {jet_ok}')
print(f'Elapsed time: {elapsed:.2f} s')
if failed:
    print('First failures:', failed[:10])
else:
    print('No discrepancies found.')
```

## Optional faster sanity check

Use this cell if the `T_1000` rational run is too slow. It does not replace the exact rational test, but it is useful for quick debugging.

```python
# Optional quick check at lower order.
QUICK_MAX_T = 100
QUICK_TRIALS = 1

quick_status = jet.projective_jet_check(
    max_index=QUICK_MAX_T,
    trials=QUICK_TRIALS,
    seed=SEED,
    verbose=False,
)
print(f'Quick projective jet check through T_{QUICK_MAX_T}: {all(quick_status.values())}')
```
