
# Projective consistency hierarchy checks

This notebook validates the closed Bell-polynomial expression for the leading-order scalar-tensor consistency hierarchy in two independent ways:

1. An exact symbolic check of the closed form against direct Riccati recursion up to a moderate order.
2. A fast exact rational jet check to much higher order, avoiding expression expansion.

The notation follows the paper: \(T_1=n_t\), \(T_2=\alpha_t\), \(T_3=\beta_t\), etc., while \(S_1=n_s-1\), \(S_2=\alpha_s\), \(S_3=\beta_s\), etc.


```python

from projective_consistency_hierarchy import ConsistencyHierarchy
import sympy as sp
```


## Exact symbolic check

The first test constructs the tensor hierarchy in two ways:

- by direct recursion from the Riccati flow \(T_1'=T_1(T_1-S_1)\);
- by the closed Bell-polynomial/Faà di Bruno expression.

We then check that the two expressions agree identically. Fully expanded symbolic expressions grow rapidly, so this check is meant for moderate order.


```python

MAX_EXACT_INDEX = 20  # checks T_2 through T_100 exactly
H = ConsistencyHierarchy(max_order=MAX_EXACT_INDEX + 2)
exact_diffs = H.exact_check(MAX_EXACT_INDEX, simplify_method="factor")
exact_diffs
```

```python

assert all(diff == 0 for diff in exact_diffs.values())
print(f"Exact symbolic check passed through T_{MAX_EXACT_INDEX}.")
```


## First few runnings in standard notation

For readability, we display the first few runnings using \(n_s,\alpha_s,\beta_s,\gamma_s\) notation.


```python

for index, name in [(2, "alpha_t"), (3, "beta_t"), (4, "gamma_t"), (5, "T_5")]:
    print(f"{name} =")
    display(H.pretty_T(index, notation="standard"))
```


## High-order rational jet check

The second test avoids expanding enormous symbolic expressions. It compares the jets generated by:

1. the projectively linearized flow \(Z'=S_1Z-1\), followed by \(T_1=1/Z\);
2. the direct Riccati flow \(T_1'=T_1(T_1-S_1)\).

At each trial, the scalar jets and the initial tensor tilt are assigned random rational values. The equality is then checked exactly over rational numbers. This is not a formal proof by itself, but it is a strong high-order implementation check, complementary to the exact symbolic validation above and to the analytic derivation in the paper.


```python

MAX_NUMERIC_INDEX = 80  # checks T_2 through T_80 using exact rational arithmetic
TRIALS = 20
H_high = ConsistencyHierarchy(max_order=MAX_NUMERIC_INDEX + 2)
status = H_high.high_order_numeric_check(max_index=MAX_NUMERIC_INDEX, trials=TRIALS, seed=2026)
failed = [idx for idx, ok in status.items() if not ok]
print(f"High-order rational jet check through T_{MAX_NUMERIC_INDEX} with {TRIALS} trials.")
print("Failures:", failed)
assert not failed
```


## Optional: push to higher order

The cell below can be increased if desired. Since this check uses rational arithmetic on jets rather than expanded expressions, it scales much better than the exact symbolic check.


```python

# Example stress test. Increase MAX_STRESS_INDEX if desired.
MAX_STRESS_INDEX = int(1e3)
STRESS_TRIALS = 1
H_stress = ConsistencyHierarchy(max_order=MAX_STRESS_INDEX + 2)
assert H_stress.all_high_order_numeric_zero(max_index=MAX_STRESS_INDEX, trials=STRESS_TRIALS, seed=31415, verbose=True, progress_every=5)
print(f"Stress test passed through T_{MAX_STRESS_INDEX} with {STRESS_TRIALS} rational trials.")
```


## Summary

The exact symbolic check confirms the closed Bell-polynomial formula through moderate order. The rational jet check confirms the same projective construction against the Riccati flow to much higher order without relying on expanded polynomial expressions.
