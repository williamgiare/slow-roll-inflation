# Symbolic check

This notebook imports `Symbolic_test.py` and checks the fully symbolic identity between the Riccati recursion and the closed Bell-polynomial/Faà di Bruno formula.

```python
# Parameters
MAX_T = 10          # Change this. Example: 1, 8, 10, 20.
SIMPLIFY_METHOD = "factor"  # Options: factor, simplify, cancel, none
```

```python
from pathlib import Path
import importlib.util

module_path = Path('Symbolic_test.py')
if not module_path.exists():
    module_path = Path('/mnt/data/Symbolic_test.py')

spec = importlib.util.spec_from_file_location('Symbolic_test', module_path)
sym = importlib.util.module_from_spec(spec)
spec.loader.exec_module(sym)

print(f'Loaded: {module_path.resolve()}')
```

```python
H = sym.ConsistencyHierarchy(max_order=max(MAX_T, 1) + 2)

if MAX_T < 1:
    raise ValueError('MAX_T must be >= 1')

if MAX_T == 1:
    print('T_1 is n_t by definition. No non-trivial symbolic identity to check.')
    symbolic_ok = True
    diffs = {}
else:
    diffs = H.exact_check(MAX_T, simplify_method=SIMPLIFY_METHOD)
    symbolic_ok = all(diff == 0 for diff in diffs.values())
    print(f'Exact symbolic Bell-vs-Riccati check through T_{MAX_T}: {symbolic_ok}')

failed = [i for i, diff in diffs.items() if diff != 0]
if failed:
    print('Failures:', failed)
else:
    print('No symbolic discrepancies found.')
```

```python
# Optional: display the differences, useful only at low order.
for i, diff in diffs.items():
    print(f'T_{i}:', diff)
```

```python
# Optional: print the first few closed-form tensor runnings in standard notation.
for i in range(2, min(MAX_T, 5) + 1):
    print(f'T_{i} =')
    display(H.pretty_T(i, notation='standard'))
```
