# Closed-form high-order check

This notebook imports `High_order_test.py` and checks the closed Bell-polynomial/Faà di Bruno relation itself against the direct Riccati hierarchy.

```python
# Parameters for the exact rational closed-form test
MAX_T = 500
TRIALS = 1
SEED = 9876
VERBOSE = True

# Optional finite-field check. Set RUN_MODULAR_CHECK = True to use it.
RUN_MODULAR_CHECK = False
MODULAR_MAX_T = 1000
PRIMES = (1000003, 1000033, 1000037)
```

```python
from pathlib import Path
import importlib.util
import time

module_path = Path('High_order_test.py')
if not module_path.exists():
    module_path = Path('/mnt/data/High_order_test.py')

spec = importlib.util.spec_from_file_location('High_order_test', module_path)
high = importlib.util.module_from_spec(spec)
spec.loader.exec_module(high)

print(f'Loaded: {module_path.resolve()}')
```

```python
start = time.time()
status = high.closed_formula_rational_check(
    max_index=MAX_T,
    trials=TRIALS,
    seed=SEED,
    verbose=VERBOSE,
)
elapsed = time.time() - start

closed_ok = all(status.values())
failed = [i for i, ok in status.items() if not ok]

print()
print(f'Closed Bell/Faa-di-Bruno rational check through T_{MAX_T}: {closed_ok}')
print(f'Elapsed time: {elapsed:.2f} s')
if failed:
    print('First failures:', failed[:10])
else:
    print('No discrepancies found.')
```

```python
if RUN_MODULAR_CHECK:
    start = time.time()
    modular_ok = high.closed_formula_modular_check(
        max_index=MODULAR_MAX_T,
        primes=PRIMES,
        seed=2468,
        verbose=VERBOSE,
    )
    elapsed = time.time() - start
    print()
    print(f'Closed Bell/Faa-di-Bruno modular check through T_{MODULAR_MAX_T}: {modular_ok}')
    print(f'Elapsed time: {elapsed:.2f} s')
else:
    print('Modular check skipped. Set RUN_MODULAR_CHECK = True in the parameter cell to run it.')
```
