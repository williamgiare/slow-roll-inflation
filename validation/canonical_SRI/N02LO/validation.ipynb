# N2LO: recovery of published results

This case tests the complete quadratic scalar and tensor amplitudes, and the tensor tilt and first two runnings through two corrections.

The comparison targets are the pivot amplitudes in
[Auclair & Ringeval (2022), Eqs. (44) and (54)](https://arxiv.org/html/2205.12608v2),
truncated to the order of this case.
The explicitly transcribed targets are in
[`published_spectra.py`](../../common/symbolic/published_spectra.py).
They are imported only by validation: production regenerates all mode coefficients.

## What is checked

- Every scalar and tensor monomial coefficient, rather than agreement at one parameter point.
- `n=0,1,2` against the separate observable-flow/Lagrange-inversion oracle.
- Exact forward substitution of the eliminated background, SR weights and the zero-`r` limit.
- Constant-`epsilon` spectra against the exact Hankel solution, including the expected remainder scaling.
- Scalar and tensor parameter-Taylor coefficients for a nonconstant canonical background, obtained from an independent linear mode solver.

The coefficient tolerances are asserted in the shared validation code and recorded
in the report. Published expressions are comparison outputs, never calculation inputs.

## Coefficient comparisons and formal identities

The scalar and tensor amplitudes are compared monomial by monomial with the independently transcribed published expressions. The consistency oracle uses a separate observable-flow/Lagrange construction.

```python
# WG Edit: show residuals without printing unwieldy generated formulas.
for row in report["checks"]["published_spectra"]["comparisons"]:
    print(f"{row['field']}: scaled coefficient residual "
          f"{row['maximum_scaled_coefficient_difference']:.3e}")
for row in report["checks"]["consistency"]:
    print(f"n={row['n']}, p={row['p']}, W={row['W']}: "
          f"exact forward-substitution residual {row['exact_forward_substitution_residual']}")
errors = [max(row["relative_coefficient_differences"])
          for row in report["checks"]["physical_taylor"]["comparisons"]]
print(f"Independent physical Taylor coefficients: max relative residual {max(errors):.3e}")
```
