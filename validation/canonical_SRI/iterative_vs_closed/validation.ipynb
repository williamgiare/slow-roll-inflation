Compare the finite-order formulas for primordial spectra and consistency relations.
Both calculations use the same numerical mode moments. The comparison checks
independent algebraic constructions as exact rational polynomials, before rounding.

## Order-by-order comparisons

Test $P_s$ and $P_t$ through $p=4$, consistency relations for $n=0,1,2,3$
through $p=4$, and LO through $n=8$. Each case compares the intermediate
background, derivatives and observable inverse, not only the displayed result.
The closed relation is also substituted back into the forward observable map.


```python
# WG Edit: execute all exact comparisons; a nonzero residual raises an error.
report = run(max_p=4, max_n=3, max_lo_n=8)
assert report["status"] == "passed"
```

## Primordial spectra

```python
for case in report["spectra"]:
    print(f"p={case['p']}: scalar/tensor spectrum residual = {case['public_spectra_residual']}")
```

## Consistency relations

```python
for case in report["consistency"]:
    print(f"n={case['n']}, p={case['p']}, W={case['W']}: "
          f"relation residual = {case['observable_relation_residual']}; "
          f"forward substitution residual = {case['forward_substitution_residual']}")
```

## Interpretation

Zero residuals establish equivalence of the two algebraic constructions on the
tested orders. They do not independently test the common moment generator,
and are not a proof of equality at untested orders. The methods accept arbitrary
finite orders; their defining identities, rather than this finite test range,
provide the general construction.
