## Select the result

`type` is `"canonical"` (the default) or `"kinflation"`.

```python
p = 2
n = 0
type = "canonical"
```

## Scalar and tensor spectra

Both expressions include their physical normalizations. The correction factors
are also available as `power.scalar_correction` and `power.tensor_correction`.

```python
power = get.spectra(p=p, type=type)
show_spectra(power)
print(power.metadata)
```

## Tensor consistency relation

The expression is $\alpha_n^t$ through $W=n+1+p$; the formal remainder starts
at weight $W+1$. The individual weight contributions are in `relation.components`.

```python
relation = get.consistency(n=n, p=p, type=type)
show_consistency(relation)
print("n, p, W:", relation.n, relation.p, relation.W)
```

## Variable scalar sound speed

The same functions select k-inflation through `type`. The tensor spectrum and
relation are evaluated at the common tensor reference. Sound derivatives use
`alpha_j^cs`, with `alpha_0^cs = d ln(c_s) / d ln(k)`.

```python
power_cs = get.spectra(p=p, type="kinflation")
relation_cs = get.consistency(n=n, p=p, type="kinflation")
show_consistency(relation_cs)
```

## Substitute observable values or export LaTeX

These operations manipulate the loaded SymPy expression. They do not rerun the
calculation. The values below are an illustrative choice of observables.

## Available orders

Both models have spectra at `p=0,...,10`. Stored relations have `n=0,...,3`
for `p=0,...,6` and `n=0` for `p=7,...,10`. Requests outside these ranges
report the missing result without starting a calculation.

Use `get.clear_cache()` to release loaded expressions. New calculations and
larger orders are provided by the separate `sri` API.
